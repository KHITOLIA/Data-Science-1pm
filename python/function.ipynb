```python
# st = "programming is awesome".lower()
# st = st.split()
def score_of_string(st):
    st = st.lower()
    st = st.split()
    v = ['a', 'e', 'i', 'o', 'u']
    total_score = 0
    for item in st:
        score = 0
        count_vowel = 0
        for char in item:
            if char in v:
                count_vowel += 1
        if count_vowel % 2 == 0:
            score = 1
        else:
            score = 0
        total_score += score
    return total_score
```

```python
score_of_string(st = "Training basket Noida aeio")
```
