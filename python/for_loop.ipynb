```python
s = input("Enter the string: ")
vow = ['a', 'e', 'i', 'o', 'u']
count_of_vowels = 0
s = s.lower()
for char in s:
    if char in vow:
        # print(char)
        count_of_vowels = count_of_vowels + 1
print(f"No of vowels in {s}: {count_of_vowels}")

```

```python
l1 = [5, 5, 1, 0,0, 1, 3, 2, 7]
l2 = []

for item in l1:
    if item not in l2:
        l2.append(item)
for i in l1:
    if i in l2:
        print(i)
```

```python
a = [12,22,22,13,12,34,34,55,67,66,55,78]
print("Duplicate elements are:")

for i in range(len(a)):
    for j in range(i + 1, len(a)):
        if a[i] == a[j]:
            print(a[i])
```
