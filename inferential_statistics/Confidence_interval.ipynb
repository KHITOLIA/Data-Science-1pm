```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Admin\Desktop\projects\Data Science-1pm\Pandas\insurance.csv")
df.head()
```

```python
pop_charges = df['charges']
```

```python
n_samples = 100
sample_size = 50
samples = []
for i in range(n_samples):
    samples.append(pop_charges.sample(sample_size).mean())
```

```python
samples
```

```python
import seaborn as sns
```

```python
sns.kdeplot(samples)
```

```python
point_estimate = np.mean(np.array(samples))
```

```python
lower_bound = point_estimate - (1.96 * np.std(pop_charges))/np.sqrt(sample_size)
upper_bound = point_estimate + (1.96 * np.std(pop_charges))/np.sqrt(sample_size)
```

```python
print(f"Confidence Interval : ({lower_bound} - {upper_bound})")
```
