```python
import kagglehub
import os
import pandas as pd
import numpy as np

# Download latest version
path = kagglehub.dataset_download("abbas829/healthcare-patient-analytics-dataset")

print("Path to dataset files:", path)
df = pd.read_csv(path + "/" + os.listdir(path)[0])
df.head()
```

```python
sns.boxplot(data = df, x = 'gender', y = 'recovery_score')
```

```python
q1 = np.quantile(df['recovery_score'],0.25)
q2 = np.median(df['recovery_score'])
q3 = np.quantile(df['recovery_score'], 0.75)
min = q1 - 1.5 * (q3 - q1)
max = q3 + 1.5 * (q3 - q1)
```

```python
a = df[df['recovery_score'] > min]
b = df[df['recovery_score'] < max]
```
