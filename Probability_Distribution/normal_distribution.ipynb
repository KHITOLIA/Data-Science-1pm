```python
import numpy as np
```

```python
age = np.random.randint(10, 100, size = 10)
```

```python
age
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
sns.kdeplot(age)
```

```python
u_age = np.mean(age)
std_age = np.std(age)
```

```python
u_age
```

```python
std_age
```

```python
new_age = (age - u_age)/std_age
```

```python
new_age
```

```python
sns.kdeplot(new_age)
```

```python
np.round(np.mean(new_age))
```

```python
np.round(np.std(new_age))
```
