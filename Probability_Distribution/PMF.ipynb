```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
l = []
for i in range(10):
    l.append(np.random.randint(1, 6))
```

```python
l = pd.Series(l)
```

```python
df = l.value_counts()/len(l)
```

```python
df.plot(kind = 'bar')
```

```python
np.cumsum(df).plot(kind = 'bar')
```

```python
l = []
for i in range(1000):
    a = np.random.randint(1, 7)
    b = np.random.randint(1, 7)
    l.append(a + b)
```

```python
l = pd.Series(l)
(l.value_counts()/len(l)).plot(kind = 'bar')
```

```python
np.cumsum(l.value_counts()/len(l)).plot(kind = 'bar')
```
