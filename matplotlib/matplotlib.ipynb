```python
def analyse_batsman(data):
    idx = 1
    plt.figure(figsize=(10,  15))
    for col in data.columns[1:]:
        plt.subplot(2, 2, idx)
        plt.scatter(data['batter'], data[col])
        plt.title(f"Batsman and their {col}")
        plt.xticks(rotation = 45)
        idx += 1
    plt.show()
```
