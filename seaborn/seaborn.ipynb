```python
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', hue = 'sex', style = 'time', size = 'size')
# plt.scatter(tips['total_bill'], tips['tip'])
plt.title("Total bill vs Tips")
plt.show()
```
